# Problem Statement

```
Mentors' Education's marketing and sales teams need visibility into which lead sources (channels), branches, and counselors are most effective at converting inquiries into enrollments, in order to allocate limited marketing budget and staff resources more efficiently. This analysis examines lead and conversion data across channels, branches, and counselors to identify high- and low-performing segments, uncover bottlenecks in the lead-to-enrollment funnel, and surface actionable, data-driven recommendations that support better resource allocation and process improvement across the organization.

```


# Summary & Key Insights (based on the analysis)

```
Overall Performance
Out of 1,200 leads generated between January 2025 and July 2026, 179 converted into enrollments — an overall conversion rate of 14.9%, while roughly a third of all leads (30.9%) were ultimately lost. This leaves meaningful room for improvement, particularly in how leads are sourced and followed up on.

```
```
Channel Performance
Referral is by far the strongest-performing channel, converting at 30.5% — more than double the next-best channel and nearly three times the rate of Facebook Ads (10.3%), which is also the single largest source of leads by volume (361 leads). This creates a clear volume-versus-quality gap: the channel bringing in the most leads is also the least efficient at converting them. Grouping channels into paid versus organic sources reinforces this pattern — organic channels convert at 18.6% compared to 10.9% for paid channels, suggesting that current ad spend may not be delivering proportionate returns. Reallocating a portion of the Facebook Ads budget toward referral incentive programs or organic lead generation could improve overall efficiency without necessarily increasing total spend.

```
```
Branch Performance
Chattogram leads in conversion rate (16.9%) despite being a mid-sized branch, while Kalabagan — the largest branch by lead volume (431 leads) — converts at a comparatively modest 16.0% and carries the highest lost-lead rate (33.2%). Sylhet, the smallest branch, has both the lowest conversion rate (11.7%) and the fastest average cycle time (29.7 days), indicating a smaller but faster-moving pipeline. These differences suggest branch-level factors — likely counselor capacity, lead quality, or local follow-up practices — are driving performance gaps that warrant closer investigation.

```
```
Counselor Performance
Individual counselor performance varies meaningfully, with conversion rates ranging from 12.0% to 19.8% across the team. Mim and Tanvir stand out as top performers, converting nearly 20% and 18% of their leads respectively, while Rafi and Nusrat trail at around 12%. This spread points to an opportunity for structured knowledge-sharing — understanding what top performers do differently in follow-up cadence or communication could help lift performance across the wider team.

```
```
Demand & Trend Patterns
The UK remains the most sought-after destination (354 leads), followed by Canada and Australia, giving a clear signal for where counselor expertise and marketing content should be concentrated. Monthly trends show noticeable volatility in conversion rate (ranging from roughly 9% to 23% month-to-month) without a strong seasonal pattern, suggesting conversion is currently driven more by lead quality and follow-up execution than by time of year.

```
```
Recommendations

```
1. Shift incremental marketing budget from Facebook Ads toward referral-driving initiatives, given the near 3x gap in conversion efficiency.
2. Investigate Kalabagan's high lost-lead rate despite strong lead volume — likely a follow-up capacity or lead-quality issue rather than a demand issue.
3. Formalize best practices from top-converting counselors (Mim, Tanvir) into training material for the wider counseling team.
4. Continue prioritizing UK, Canada, and Australia-focused resources, given they account for over two-thirds of total demand.

#### Outlier `Detection` & `Removal`



````python
df.describe()
````

````python
df[df.follow_up_count<0]
````

````python
df['cycle_days'] = (df.last_update_date - df.created_date).dt.days
df[df.cycle_days<0]
````

````python
df.info()
````

````python
cols = ['follow_up_count', 'cycle_days']
for i in cols:
    mean = df[i].mean()
    std = df[i].std()
    z_score = (df[i] - mean) / std
    df[f'{i}_z_score'] = z_score
df.head()
````

````python
outlier_profile = df[(((df['follow_up_count_z_score'] > 3) | (df['follow_up_count_z_score'] > 3)) | ((df['cycle_days_z_score'] > 3) | (df['cycle_days_z_score'] > 3)))]
````

````python
follow_max = outlier_profile.follow_up_count.max()
follow_min = outlier_profile.follow_up_count.min()
cycle_max = outlier_profile.cycle_days.max()
cycle_min = outlier_profile.cycle_days.min()
print(f"Follow up count: {follow_min} to {follow_max}")
print(f"Cycle days: {cycle_min} to {cycle_max}")
````
